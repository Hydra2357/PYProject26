```python
def equilibriun_index(arr):
    total = sum(arr)
    left_sum = 0

    for i in range(len(arr)):
        right_sum = total - left_sum - arr[i]

        if left_sum == right_sum:
            return i
        left_sum += arr[i]

    return -1

l1 = [1, 3, 5, 2 ,2]
print(equilibriun_index(l1))
print(equilibriun_index([1, 2, 5 ,6, 7]))
```

```python
def find_first(arr, target):
    low, high = 0, len(arr)-1
    result = -1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == target:
            result = mid
            high = mid - 1

        elif arr[mid] < target:
            low = mid + 1

        else:
            high = mid - 1
    return result

def find_last(arr, target):
    low, high = 0, len(arr) -1
    result = -1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == target:
            result = mid
            low = mid + 1

        elif arr[mid] < target:
            low = mid + 1

        else:
            high = mid - 1

    return result

def first_last_occurrence(arr, target):
    return find_first(arr, target), find_last(arr, target)

print(first_last_occurrence([1,2,2,2,3,4], 2))
print(first_last_occurrence([1,2,2,2,3,4], 4))
print(first_last_occurrence([1,2,2,2,3,4], 5))
print(first_last_occurrence([1,2,2,2,3,4], 3))
```

```python
def count_arrangements(N, R, end):
    if N == 0:
        return 0
    if N == 1:
        return 1 if end == 1 else 0   # Must start and end with 1

    # dp[i][j] = number of valid ways to arrange i soldiers ending with soldier j
    dp = [[0] * (R + 1) for _ in range(N + 1)]

    # First position must be 1
    dp[1][1] = 1

    for i in range(2, N + 1):
        for prev in range(1, R + 1):
            if dp[i-1][prev] == 0:
                continue
            for curr in range(1, R + 1):
                if curr != prev:                    # No two adjacent same
                    dp[i][curr] += dp[i-1][prev]

    return dp[N][end]


# Input
N, R, end = map(int, input().split())

result = count_arrangements(N, R, end)
print(result)
```

```python
N, R, end = map(int, input().split())
# Stores the number of valid sequences
count = 0

# Creates an array of size N to build sequences
# Initially filled with 0 → placeholder
arr = [0] * N

# pos → current index being filled
# count is global so it can be updated inside recursion
def generate(pos):
    global count

# If we filled all positions -> valid seq formed
# increment count
    if pos == N:
        count += 1
        return

    if pos == N-1:
        if arr[pos-1] != end:
            arr[pos] = end
            generate(pos + 1)
        return

# Normal Case (Middle Positions)
    for i in range(1, R+1):
        if pos > 0 and arr[pos-1] == i:
            continue
        arr[pos] = i
        generate(pos + 1)

arr[0] = 1
generate(1)

print(count)
```

```python
def max_subset_sum_optimized(arr, maxSum):
    if maxSum <= 0 or not arr:
        return 0

    # 1D DP
    dp = [0] * (maxSum + 1)

    for num in arr:
        # Traverse backward to avoid using same element multiple times
        for j in range(maxSum, num - 1, -1):
            dp[j] = max(dp[j], dp[j - num] + num)

    # Find the maximum value in dp[0..maxSum]
    return max(dp)


# Example
if __name__ == "__main__":
    arr = [1, 2, 3, 4]
    maxSum = 6
    print(max_subset_sum_optimized(arr, maxSum))  # Output: 6
```

```python
def min_adjacent_swaps(S, X):
    swaps = 0
    n = len(S)

    for i in range(n):
        if S[i] != X[i]:
            j = i + 1

            while j < n and S[j] != X[i]:
                j += 1

            if j == n:
                return -1   # Not possible

            while j > i:
                S[j], S[j-1] = S[j-1], S[j]
                swaps += 1
                j -= 1
    return swaps

N = int(input())
s = list(map(int, input().split()))
x = list(map(int, input().split()))

print(min_adjacent_swaps(s, x))
```

```python
# find the elements that appears in more than N/2 time. if none, return -1
def majorityElement(arr):
    if not arr:
        return -1

    candidate = None
    count = 0

    for num in arr:
        if count == 0:
            candidate = num
        count += (1 if num == candidate else -1)

    if arr.count(candidate) > len(arr) // 2:
        return candidate
    return -1

print(majorityElement([3, 2, 3]))
print(majorityElement([6,7,5,6,7]))
print(majorityElement([2, 2, 1, 1, 2]))

# TC = O(n)
# SC = O(1)
```

```python
def twoSorted(arr, target):
    left, right = 0, len(arr) -1

    while left < right:
        currentSum = arr[left] + arr[right]

        if currentSum == target:
            return [left, right]
        elif currentSum < target:
            left += 1
        else:
            right -= 1
    return [-1, -1]

print(twoSorted([3, 2, 3], 5))
print(twoSorted([6, 7, 5, 6, 7], 6))
print(twoSorted([2, 2, 1, 1, 2], 111))
```

```python
def transposeMatrix(matrix):
    if not matrix or not matrix[0]:
        return []

    rows = len(matrix)
    col = len(matrix[0])

    transpose = [[0] * rows for _ in range(col)]

    for i in range(rows):
        for j in range(col):
            transpose[j][i] = matrix[i][j]

    return transpose

matrix = [
    [1, 2, 3],
    [4, 5, 6],
]
resultnew = transposeMatrix(matrix)
print("This is transpose matrix: ")
for row in resultnew:
    print(row)
```

```python
def print_matrix(matrix, name="Matrix"):
    """Helper function to print a matrix nicely"""
    print(f"\n{name}:")
    for row in matrix:
        print("  ", end="")
        for element in row:
            print(f"{element:8}", end=" ")
        print()

def get_matrix_input(rows, cols, name="Matrix"):
    """Get matrix input from user"""
    print(f"\nEnter elements for {name} ({rows} x {cols}):")
    matrix = []
    for i in range(rows):
        while True:
            try:
                row = list(map(int, input(f"Row {i+1} ({cols} elements): ").strip().split()))
                if len(row) != cols:
                    print(f"Error: Please enter exactly {cols} elements.")
                    continue
                matrix.append(row)
                break
            except ValueError:
                print("Error: Please enter only integers.")
    return matrix

def are_compatible_add_sub(A, B):
    """Check if two matrices are compatible for addition or subtraction"""
    return len(A) == len(B) and len(A[0]) == len(B[0])

def are_compatible_multiply(A, B):
    """Check if two matrices are compatible for multiplication"""
    return len(A[0]) == len(B)

def add_matrices(A, B):
    """Add two matrices"""
    if not are_compatible_add_sub(A, B):
        raise ValueError("Matrices must have the same dimensions for addition.")

    rows, cols = len(A), len(A[0])
    result = [[0 for _ in range(cols)] for _ in range(rows)]

    for i in range(rows):
        for j in range(cols):
            result[i][j] = A[i][j] + B[i][j]
    return result

def subtract_matrices(A, B):
    """Subtract matrix B from matrix A"""
    if not are_compatible_add_sub(A, B):
        raise ValueError("Matrices must have the same dimensions for subtraction.")

    rows, cols = len(A), len(A[0])
    result = [[0 for _ in range(cols)] for _ in range(rows)]

    for i in range(rows):
        for j in range(cols):
            result[i][j] = A[i][j] - B[i][j]
    return result

def multiply_matrices(A, B):
    """Multiply two matrices"""
    if not are_compatible_multiply(A, B):
        raise ValueError(f"Cannot multiply: Columns of first matrix ({len(A[0])}) "
                        f"must equal rows of second matrix ({len(B)}).")

    rows_A, cols_A = len(A), len(A[0])
    rows_B, cols_B = len(B), len(B[0])

    result = [[0 for _ in range(cols_B)] for _ in range(rows_A)]

    for i in range(rows_A):
        for j in range(cols_B):
            for k in range(cols_A):
                result[i][j] += A[i][k] * B[k][j]
    return result

def main():
    print("=== MATRIX OPERATIONS PROGRAM ===\n")

    while True:
        print("\nChoose operation:")
        print("1. Matrix Addition")
        print("2. Matrix Subtraction")
        print("3. Matrix Multiplication")
        print("4. Exit")

        choice = input("\nEnter your choice (1-4): ").strip()

        if choice == '4':
            print("Thank you for using Matrix Operations Program!")
            break

        if choice not in ['1', '2', '3']:
            print("Invalid choice! Please enter 1, 2, 3, or 4.")
            continue

        try:
            # Get dimensions
            if choice in ['1', '2']:
                rows = int(input("Enter number of rows: "))
                cols = int(input("Enter number of columns: "))

                print("\n--- Enter First Matrix ---")
                A = get_matrix_input(rows, cols, "Matrix A")

                print("\n--- Enter Second Matrix ---")
                B = get_matrix_input(rows, cols, "Matrix B")

                print_matrix(A, "Matrix A")
                print_matrix(B, "Matrix B")

                if choice == '1':
                    result = add_matrices(A, B)
                    print_matrix(result, "Result (A + B)")
                else:
                    result = subtract_matrices(A, B)
                    print_matrix(result, "Result (A - B)")

            else:  # Multiplication
                # Get dimensions for Matrix A
                rows_A = int(input("Enter rows for Matrix A: "))
                cols_A = int(input("Enter columns for Matrix A: "))

                print("\n--- Enter First Matrix (A) ---")
                A = get_matrix_input(rows_A, cols_A, "Matrix A")

                # Get dimensions for Matrix B
                rows_B = cols_A  # Must match for multiplication
                cols_B = int(input(f"\nEnter columns for Matrix B: "))

                print("\n--- Enter Second Matrix (B) ---")
                B = get_matrix_input(rows_B, cols_B, "Matrix B")

                print_matrix(A, "Matrix A")
                print_matrix(B, "Matrix B")

                result = multiply_matrices(A, B)
                print_matrix(result, "Result (A × B)")

        except ValueError as e:
            print(f"\nError: {e}")
        except Exception as e:
            print(f"\nUnexpected error: {e}")

        print("\n" + "="*50)

if __name__ == "__main__":
    main()
```

```python
def matrix_operations(A, B):
    n = len(A)
    m = len(A[0])
    n2 = len(B)
    m2 = len(B[0])

    if n == n2 and m == m2:
        add = [[0]*m for _ in range(n)]
        sub = [[0]*m for _ in range(n)]

        for i in range(n):
            for j in range(m):
                add[i][j] = A[i][j] + B[i][j]
                sub[i][j] = A[i][j] - B[i][j]
    else:
        add = "Addition not possible (different dimensions)"
        sub = "Subtraction not possible (different dimensions)"

    if m == n2:
        mul = [[0]*m2 for _ in range(n)]

        for i in range(n):
            for j in range(m2):
                for k in range(m):
                    mul[i][j] += A[i][k] * B[k][j]
    else:
        mul = "Multiplication not possible (incompatible dimensions)"

    return add, sub, mul

A = [
    [1, 2, 3],
    [4, 5, 6]
]

B = [
    [7, 8, 9],
    [1, 2, 3]
]

add, sub, mul = matrix_operations(A, B)

print("Addition:")
print(add)

print("\nSubtraction:")
print(sub)

print("\nMultiplication:")
print(mul)
```

```python
def add_matrix(A, B):
    n, m = len(A), len(A[0])

    if n != len(B) or m != len(B[0]):
        return "Addition not possible (different dimensions)"

    result = [[0]*m for _ in range(n)]

    for i in range(n):
        for j in range(m):
            result[i][j] = A[i][j] + B[i][j]

    return result

def subtract_matrix(A, B):
    n, m = len(A), len(A[0])

    if n != len(B) or m != len(B[0]):
        return "Subtraction not possible (different dimensions)"

    result = [[0]*m for _ in range(n)]

    for i in range(n):
        for j in range(m):
            result[i][j] = A[i][j] - B[i][j]

    return result

def multiply_matrix(A, B):
    n, m = len(A), len(A[0])
    n2, m2 = len(B), len(B[0])

    if m != n2:
        return "Multiplication not possible (incompatible dimensions)"

    result = [[0]*m2 for _ in range(n)]

    for i in range(n):
        for j in range(m2):
            for k in range(m):
                result[i][j] += A[i][k] * B[k][j]

    return result

if __name__ == "__main__":
    A = [
        [1, 2, 3],
        [4, 5, 6]
    ]

    B = [
        [7, 8, 9],
        [1, 2, 3]
    ]

    print("Addition:")
    print(add_matrix(A, B))

    print("\nSubtraction:")
    print(subtract_matrix(A, B))

    print("\nMultiplication:")
    print(multiply_matrix(A, B))
```

```python
def transaction_monitor():
    n = int(input())

    seen = set()
    prev_time = None

    for _ in range(n):
        sender, receiver, timestamp, amount = input().split()
        timestamp = int(timestamp)

        if (sender, receiver) in seen:
            print("Error: Duplicate Transaction")
            return

        seen.add((sender, receiver))

        if prev_time is not None:
            if timestamp - prev_time > 60:
                print("Fraud Detected")
                return

        prev_time = timestamp
    print("All Transactions Valid")

if __name__ == "__main__":
    transaction_monitor()
```
