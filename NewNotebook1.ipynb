```python
class PowerOfTwo:
    """Custom iterator to find power of 2 upto a max exponent."""
    def __init__(self, max_exponent):
        self.max_exponent = max_exponent
        self.current = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= self.max_exponent:
            result = 2 ** self.current
            self.current +=1
            return result
        else:
            raise StopIteration

pow_iter = PowerOfTwo(3)
for val in pow_iter:
    print(val)
```

```python
class CountDown:
    def __init__(self, start):
        self.current = start

    def __iter__(self):
        return self

    def __next__(self):
        if self.current < 0:
            raise StopIteration

        value = self.current
        self.current -= 1
        return value

for i in CountDown(6):
    print(i)
```

```python
def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a+b

for num in fibonacci(10):
    print(num, end=" ")

def read_large_file(filename):
    with open(filename, 'r') as file:
        for line in file:
            yield line.strip()
```

```python
#news app code
import tkinter as tk
from tkinter import ttk, messagebox

NEWS_DATA = [
    {
        'title': 'AI Revolution in 2026',
        'category': 'Technology',
        'description': 'Artificial intelligence is transforming industries worldwide.'
    },
    {
        'title': 'Global Markets Rally',
        'category': 'Business',
        'description': 'Stock markets showed strong gains amid positive economic data.'
    },
    {
        'title': 'New Space Mission Announced',
        'category': 'Science',
        'description': 'A new international mission aims to explore deep space.'
    },
    {
        'title': 'Championship Finals Tonight',
        'category': 'Sports',
        'description': 'Fans are eagerly awaiting the season finale.'
    }
]

class NewsApp:
    def __init__(self, root):
        self.root = root
        self.root.title('News App')
        self.root.geometry('700x500')

        self.search_var = tk.StringVar()

        header = tk.Label(
            root,
            text='News App',
            font=('Arial', 24, 'bold')
        )
        header.pack(pady=10)

        search_frame = tk.Frame(root)
        search_frame.pack(fill='x', padx=20, pady=10)

        search_entry = ttk.Entry(
            search_frame,
            textvariable=self.search_var,
            font=('Arial', 12)
        )
        search_entry.pack(side='left', fill='x', expand=True, padx=(0, 10))

        search_button = ttk.Button(
            search_frame,
            text='Search',
            command=self.search_news
        )
        search_button.pack(side='left')

        self.news_listbox = tk.Listbox(root, font=('Arial', 12), height=12)
        self.news_listbox.pack(fill='both', expand=True, padx=20, pady=10)
        self.news_listbox.bind('<<ListboxSelect>>', self.show_news_details)

        self.load_news(NEWS_DATA)

    def load_news(self, news_items):
        self.news_listbox.delete(0, tk.END)
        self.current_news = news_items

        for article in news_items:
            self.news_listbox.insert(
                tk.END,
                f"[{article['category']}] {article['title']}"
            )

    def search_news(self):
        query = self.search_var.get().lower().strip()

        if not query:
            self.load_news(NEWS_DATA)
            return

        filtered_news = [
            article for article in NEWS_DATA
            if query in article['title'].lower()
            or query in article['category'].lower()
            or query in article['description'].lower()
        ]

        if filtered_news:
            self.load_news(filtered_news)
        else:
            messagebox.showinfo('No Results', 'No matching news articles found.')

    def show_news_details(self, event):
        selection = self.news_listbox.curselection()
        if not selection:
            return

        article = self.current_news[selection[0]]
        details = (
            f"Title: {article['title']}\n\n"
            f"Category: {article['category']}\n\n"
            f"Description: {article['description']}"
        )

        messagebox.showinfo('Article Details', details)


if __name__ == '__main__':
    root = tk.Tk()
    app = NewsApp(root)
    root.mainloop()
```

```python
import re

text = """
Name: Alice
Email: alice123@gmail.com
Phone: 9876543210
Date: 30-04-2026
Price: $499
Weight: 25kg
Languages: Python, Java, C++
"""

print("=" * 60)
print("ORIGINAL TEXT")
print("=" * 60)
print(text)

# ---------------------------------------------------
# 1. Character Classes
# ---------------------------------------------------
print("\n1. Character Classes")
print(re.findall(r"[A-Z][a-z]+", text))   # Words starting with capital letters

# ---------------------------------------------------
# 2. Predefined Character Classes
# ---------------------------------------------------
print("\n2. Predefined Character Classes")
print("Digits:", re.findall(r"\d", text))
print("Words:", re.findall(r"\w+", text))
print("Whitespace count:", len(re.findall(r"\s", text)))

# ---------------------------------------------------
# 3. Quantifiers
# ---------------------------------------------------
print("\n3. Quantifiers")
print("One or more digits:", re.findall(r"\d+", text))
print("Exactly 10 digits:", re.findall(r"\d{10}", text))
print("2 to 4 digits:", re.findall(r"\d{2,4}", text))

# ---------------------------------------------------
# 4. Anchors
# ---------------------------------------------------
print("\n4. Anchors")
print("Starts with 'Name':", bool(re.search(r"^Name", text, re.MULTILINE)))
print("Ends with digits:", re.findall(r"\d+$", text, re.MULTILINE))

# ---------------------------------------------------
# 5. Word Boundary
# ---------------------------------------------------
print("\n5. Word Boundary")
print(re.findall(r"\bPython\b", text))

# ---------------------------------------------------
# 6. Grouping and Alternation
# ---------------------------------------------------
print("\n6. Grouping and Alternation")
print(re.findall(r"(Python|Java|C\+\+)", text))

# ---------------------------------------------------
# 7. Named Groups
# ---------------------------------------------------
print("\n7. Named Groups")
email_match = re.search(
    r"(?P<username>[\w.]+)@(?P<domain>[\w.]+)",
    text
)
if email_match:
    print("Username:", email_match.group("username"))
    print("Domain:", email_match.group("domain"))

# ---------------------------------------------------
# 8. Lookahead
# ---------------------------------------------------
print("\n8. Positive Lookahead")
print(re.findall(r"\d+(?=kg)", text))   # Number before 'kg'

# ---------------------------------------------------
# 9. Lookbehind
# ---------------------------------------------------
print("\n9. Positive Lookbehind")
print(re.findall(r"(?<=\$)\d+", text))  # Number after '$'

# ---------------------------------------------------
# 10. Greedy vs Non-Greedy
# ---------------------------------------------------
html = "<h1>Title</h1><p>Paragraph</p>"
print("\n10. Greedy vs Non-Greedy")
print("Greedy:", re.findall(r"<.*>", html))
print("Non-Greedy:", re.findall(r"<.*?>", html))

# ---------------------------------------------------
# 11. Substitution
# ---------------------------------------------------
print("\n11. Substitution")
print(re.sub(r"\d", "*", text))

# ---------------------------------------------------
# 12. Split
# ---------------------------------------------------
print("\n12. Split")
print(re.split(r"[:,\n]+", text))

# ---------------------------------------------------
# 13. Match / Search / Fullmatch
# ---------------------------------------------------
print("\n13. Match / Search / Fullmatch")
print("match():", re.match(r"Name", text.strip()))
print("search():", re.search(r"Java", text))
print("fullmatch():", re.fullmatch(r"\d{10}", "9876543210"))

# ---------------------------------------------------
# 14. Finditer
# ---------------------------------------------------
print("\n14. Finditer")
for match in re.finditer(r"\d+", text):
    print(f"Found '{match.group()}' at {match.span()}")

# ---------------------------------------------------
# 15. Flags
# ---------------------------------------------------
print("\n15. Flags")
print(re.findall(r"python", text, re.IGNORECASE))

print("\n" + "=" * 60)
print("REGEX CHEAT SHEET COMPLETE")
print("=" * 60)
```

#### AsyncIO in python


asyncio is Python's built-in library for writing concurrent code using the async/await syntax. It is ideal for I/O-bound tasks (network requests, file operations, databases, APIs) but not for CPU-bound work (use multiprocessing or concurrent.futures for that).
1. Core Concepts

Coroutine: A function declared with async def. It can be paused and resumed.
Event Loop: The core scheduler that runs coroutines, handles I/O, and manages callbacks. Usually accessed via asyncio.get_running_loop().
Task: A coroutine wrapped in a Task object so it can run concurrently.
Future: A low-level object representing a result that will be available later.
await: Pauses the current coroutine until the awaited thing completes.

```python
import asyncio

async def say_hello():
    print("Hello")
    await asyncio.sleep(1)
    print("World")

await say_hello()
```

```python
import asyncio
import time

async def worker(name, delay):
    print(f"{name} starting")
    await asyncio.sleep(delay)
    print(f"{name} finished")
    return f"{name} result"

async def main():
    start = time.time()

    # Run sequentially
    # await worker("A", 1)
    # await worker("B", 2)

    # Run concurrently
    task1 = asyncio.create_task(worker("A", 1))
    task2 = asyncio.create_task(worker("B", 2))

    results = await asyncio.gather(task1, task2)
    print("Results:", results)

    print("Total time:", time.time() - start)

await main()
```

```python
import asyncio
'''Pattern 1: Create and await individually'''
async def function1():
    await asyncio.sleep(1)
    print("Func 1 done")
    return 10

async def function2():
    await asyncio.sleep(1)
    print("Func 2 done")
    return 20

async def main():
    # Create tasks
    task1 = asyncio.create_task(function1())
    task2 = asyncio.create_task(function2())

    # Await them individually
    result1 = await task1
    result2 = await task2
    print(f"Results: {result1}, {result2}")

await main()
```

```python
"""Pattern 2: Create tasks in a loop"""
async def process_item(item):
    await asyncio.sleep(1)
    print(f"Processed {item}")
    return item * 2

async def main():
    items = [1, 2, 3, 4, 5]
    new_tasks = [asyncio.create_task(process_item(item)) for item in items]
    new_res = await asyncio.gather(*new_tasks)
    print(f"All results: {new_res}")

await main()
```

```python
"""Pattern 3: Don't await immediately (let them run in background)"""
import asyncio

async def background_task():
    for i in range(5):
        await asyncio.sleep(1)
        print(f"Background: {i+1}/5")

async def main_task():
    for i in range(3):
        await asyncio.sleep(1.5)
        print(f"Main: step {i+1}")

async def main():
    # Start background task without waiting for it
    bg_task = asyncio.create_task(background_task())

    # Run main task while background runs
    await main_task()

    # Still need to wait for background to finish if needed
    await bg_task
    print("All done!")

await main()
```

```python
async def long_task():
    try:
        for i in range(10):
            await asyncio.sleep(0.5)
            print(f"working...{i}")
    except asyncio.CancelledError:
        print("task was cancelled")
        raise
        # raise ValueError("restart the program...")

async def main2():
    t1 = asyncio.create_task(long_task())
    await asyncio.sleep(1)
    t1.cancel()

    try:
        await t1
    except asyncio.CancelledError:
        print("Main caught cancellation...")

asyncio.run(main2())
```

```python
import asyncio
async def worker(lock, name):
    async with lock:
        print(f"{name} acquired the lock")
        await asyncio.sleep(1)
        print(f"{name} releasing the lock")

async def main3():
    lock = asyncio.Lock()
    await asyncio.gather(
        worker(lock, "A"),
        worker(lock, "B"),
        worker(lock, "C")
    )

asyncio.run(main3())
```

Design Patterns in Python are reusable solutions to common software design problems. They are divided into three main categories:

Creational Patterns – Deal with object creation mechanisms.
Structural Patterns – Deal with object composition and class relationships.
Behavioral Patterns – Deal with communication and responsibility between objects.

```python
class Singleton:
    _instance = None

    def __new__(cls, *args, **kwargs):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance

    def __init__(self, value=None):
        # Prevent reinitialization
        if not hasattr(self, 'initialized'):
            self.value = value
            self.initialized = True


# Usage
s1 = Singleton("Hello")
s2 = Singleton("World")
print(s1.value)   # Hello
print(s2.value)   # Hello
print(s1 is s2)   # True
```

```python
class SIngletonMeta(type):
    _instances = {}

    def __call__(cls, *args, **kwargs):
        if cls not in cls._instances:
            cls._instances[cls] = super().__call__(*args, **kwargs)
        return cls._instances[cls]
```

```python
from typing import Any, Union, Optional, Literal, ClassVar, Final, NoReturn
from dataclasses import dataclass
import sys

# 1. ANY - Accepts any type
print("=== ANY Example ===")
def flexible_function(data: Any) -> Any:
    return f"Got: {data}"

print(flexible_function(42))
print(flexible_function("hello"))
print(flexible_function([1, 2, 3]))
print()

# 2. UNION - Variable can be multiple types
print("=== UNION Example ===")
def process_value(value: Union[int, str, float]) -> str:
    if isinstance(value, int):
        return f"Integer: {value * 2}"
    elif isinstance(value, str):
        return f"String: {value.upper()}"
    else:
        return f"Float: {value * 1.5}"

print(process_value(10))
print(process_value("hello"))
print(process_value(3.14))
print()

# 3. OPTIONAL - Can be None or specific type
print("=== OPTIONAL Example ===")
def find_user(user_id: int, database: Optional[dict] = None) -> str:
    if database is None:
        database = {1: "Alice", 2: "Bob"}
    return database.get(user_id, "User not found")

print(find_user(1))  # Uses default database
print(find_user(3))  # User not found
print(find_user(2, {2: "Charlie"}))  # Custom database
print()

# 4. LITERAL - Only specific literal values allowed
print("=== LITERAL Example ===")
from typing import Literal

def set_status(status: Literal['active', 'inactive', 'pending']) -> str:
    if status == 'active':
        return "✅ System is running"
    elif status == 'inactive':
        return "⏸️ System is paused"
    else:
        return "⏳ System is initializing"

print(set_status('active'))
print(set_status('pending'))
# print(set_status('invalid'))  # This would cause type error in type checker
print()

# 5. CLASSVAR - Class level variable
print("=== CLASSVAR Example ===")
@dataclass
class Counter:
    instance_id: int
    class_var: ClassVar[int] = 0  # Shared across all instances

    def __post_init__(self):
        Counter.class_var += 1

    @classmethod
    def get_total_instances(cls) -> int:
        return cls.class_var

obj1 = Counter(1)
obj2 = Counter(2)
obj3 = Counter(3)
print(f"Total instances created: {Counter.get_total_instances()}")
print()

# 6. FINAL - Constant value that shouldn't change
print("=== FINAL Example ===")
MAX_RETRIES: Final[int] = 3

def retry_operation(attempt: int) -> str:
    if attempt > MAX_RETRIES:
        return "Operation failed after max retries"
    return f"Attempt {attempt} of {MAX_RETRIES}"

print(retry_operation(1))
print(retry_operation(4))
# MAX_RETRIES = 5  # Type checker would flag this as error
print()

# 7. NORETURN - Function never returns (always raises exception or infinite loop)
print("=== NORETURN Example ===")
def error_exit(message: str) -> NoReturn:
    print(f"ERROR: {message}")
    sys.exit(1)

def validate_age(age: int) -> int:
    if age < 0:
        error_exit("Age cannot be negative")
    return age

try:
    result = validate_age(25)
    print(f"Valid age: {result}")
    # This line won't execute because error_exit exits
    validate_age(-5)
except SystemExit:
    print("Program terminated due to invalid age")
```

```python
from typing import List, Tuple, Dict, Set, FrozenSet, Deque, DefaultDict, OrderedDict
from collections import deque, defaultdict, Counter
import sys

# 1. LIST - Mutable sequence
print("=== LIST Example ===")
def process_scores(scores: List[int]) -> List[str]:
    result = []
    for score in scores:
        if score >= 90:
            result.append("A")
        elif score >= 80:
            result.append("B")
        elif score >= 70:
            result.append("C")
        else:
            result.append("F")
    return result

student_scores: List[int] = [95, 82, 88, 67, 91, 76]
grades = process_scores(student_scores)
print(f"Scores: {student_scores}")
print(f"Grades: {grades}")

# List with mixed types (using Union)
mixed_list: List[Union[int, str]] = [1, "hello", 2, "world", 3]
print(f"Mixed list: {mixed_list}")
print()

# 2. TUPLE - Immutable sequence
print("=== TUPLE Example ===")
def get_user_info() -> Tuple[str, int, List[str]]:
    name = "Alice Johnson"
    age = 28
    hobbies = ["reading", "coding", "hiking"]
    return (name, age, hobbies)

user_info = get_user_info()
name, age, hobbies = user_info  # Unpacking
print(f"Name: {name}")
print(f"Age: {age}")
print(f"Hobbies: {hobbies}")

# Fixed-length tuple
coordinates: Tuple[int, int] = (10, 20)
print(f"Coordinates: {coordinates}")
print()

# 3. DICT - Key-value pairs
print("=== DICT Example ===")
def update_inventory(inventory: Dict[str, int], items_sold: Dict[str, int]) -> Dict[str, int]:
    updated = inventory.copy()
    for item, quantity in items_sold.items():
        if item in updated:
            updated[item] -= quantity
            if updated[item] < 0:
                updated[item] = 0
    return updated

inventory: Dict[str, int] = {"apple": 50, "banana": 30, "orange": 25}
sold: Dict[str, int] = {"apple": 10, "banana": 5, "grape": 3}
updated_inventory = update_inventory(inventory, sold)
print(f"Original: {inventory}")
print(f"After update: {updated_inventory}")

# Nested dictionary
nested_dict: Dict[str, Dict[str, int]] = {
    "user1": {"score": 100, "level": 3},
    "user2": {"score": 85, "level": 2}
}
print(f"Nested dict: {nested_dict}")
print()

# 4. SET - Unique elements
print("=== SET Example ===")
def find_common_tags(tags1: Set[str], tags2: Set[str]) -> Set[str]:
    return tags1.intersection(tags2)

user1_tags: Set[str] = {"python", "data", "machine learning", "ai"}
user2_tags: Set[str] = {"python", "javascript", "ai", "web"}
common = find_common_tags(user1_tags, user2_tags)
print(f"User1 tags: {user1_tags}")
print(f"User2 tags: {user2_tags}")
print(f"Common tags: {common}")

# Set operations
numbers: Set[int] = {1, 2, 3, 4, 5}
numbers.add(6)
numbers.discard(3)
print(f"Numbers set: {numbers}")
print()

# 5. FROZENSET - Immutable set
print("=== FROZENSET Example ===")
def has_permission(user_permissions: FrozenSet[str], required: FrozenSet[str]) -> bool:
    return required.issubset(user_permissions)

admin_perms: FrozenSet[str] = frozenset(["read", "write", "delete", "execute"])
required_perms: FrozenSet[str] = frozenset(["read", "write"])
can_access = has_permission(admin_perms, required_perms)
print(f"Admin permissions: {admin_perms}")
print(f"Required: {required_perms}")
print(f"Access granted: {can_access}")

# FrozenSet as dictionary key (can't do with regular set)
perm_map = {
    frozenset(["read"]): "viewer",
    frozenset(["read", "write"]): "editor",
    frozenset(["read", "write", "delete"]): "admin"
}
print(f"Permission map: {perm_map[frozenset(['read', 'write'])]}")
print()

# 6. DEQUE - Double-ended queue (fast append/pop from both ends)
print("=== DEQUE Example ===")
def simulate_queue(operations: List[Tuple[str, Optional[int]]]) -> Deque[int]:
    queue: Deque[int] = deque()

    for op, value in operations:
        if op == "append" and value is not None:
            queue.append(value)
            print(f"Appended {value}: {list(queue)}")
        elif op == "appendleft" and value is not None:
            queue.appendleft(value)
            print(f"Appended left {value}: {list(queue)}")
        elif op == "pop":
            if queue:
                removed = queue.pop()
                print(f"Popped {removed}: {list(queue)}")
        elif op == "popleft":
            if queue:
                removed = queue.popleft()
                print(f"Popped left {removed}: {list(queue)}")

    return queue

ops: List[Tuple[str, Optional[int]]] = [
    ("append", 1), ("append", 2), ("appendleft", 0),
    ("pop", None), ("popleft", None), ("append", 3)
]
result_queue = simulate_queue(ops)
print()

# 7. DEFAULTDICT - Dictionary with default values
print("=== DEFAULTDICT Example ===")
def count_words(text: List[str]) -> DefaultDict[str, int]:
    word_count: DefaultDict[str, int] = defaultdict(int)

    for word in text:
        word_count[word] += 1

    return word_count

text = ["apple", "banana", "apple", "orange", "banana", "apple"]
counts = count_words(text)
print(f"Text: {text}")
print(f"Word counts: {dict(counts)}")

# DefaultDict with list default
grouped_data: DefaultDict[str, List[int]] = defaultdict(list)
for key, value in [("A", 1), ("B", 2), ("A", 3), ("B", 4), ("C", 5)]:
    grouped_data[key].append(value)
print(f"Grouped data: {dict(grouped_data)}")
print()

# 8. ORDEREDDICT - Dictionary that remembers insertion order
print("=== ORDEREDDICT Example ===")
from collections import OrderedDict

class LRUCache:
    def __init__(self, capacity: int):
        self.cache: OrderedDict[str, int] = OrderedDict()
        self.capacity = capacity

    def get(self, key: str) -> int:
        if key not in self.cache:
            return -1
        # Move to end (most recently used)
        self.cache.move_to_end(key)
        return self.cache[key]

    def put(self, key: str, value: int) -> None:
        if key in self.cache:
            self.cache.move_to_end(key)
        self.cache[key] = value
        if len(self.cache) > self.capacity:
            self.cache.popitem(last=False)  # Remove least recently used

    def display(self) -> None:
        print(f"Cache: {list(self.cache.items())}")

cache = LRUCache(3)
cache.put("A", 1)
cache.put("B", 2)
cache.put("C", 3)
cache.display()
cache.get("A")
cache.put("D", 4)
cache.display()
print()

# Bonus: COUNTER (from collections, often used with typing)
from typing import Counter

def analyze_frequency(data: List[str]) -> Counter[str]:
    return Counter(data)

items = ["red", "blue", "red", "green", "blue", "red", "yellow"]
frequency = analyze_frequency(items)
print("=== COUNTER Example ===")
print(f"Items: {items}")
print(f"Frequency: {dict(frequency)}")
print(f"Most common: {frequency.most_common(2)}")
```
