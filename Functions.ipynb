```python
def fibonacci_series(n):
    if n <= 0:
        return []
    fib = [0,1]

    for i in range(2, n):
        nextnumber = fib[-1] + fib[-2]
        fib.append(nextnumber)

    return fib[:n]

if __name__ == "__main__":
    numterms = 13
    series = fibonacci_series(numterms)

    print(f"Fibonacci series: {series}")
    print(series)

    print("formatted output:")
    for i, num in enumerate(series):
        print(f"Term {i}: {num}")
```

```python
def count_the_number(num):
    count = 0
    while num > 0:
        count += 1
        num = num // 10
    return count

print(count_the_number(435))
print(count_the_number(4))
print(count_the_number(4354))
count_the_number(4545)
```

```python
def check_palindrome(num):
    origianl = num
    result = 0

    while num > 0:
        large_digit = num % 10
        result = result * 10 + large_digit
        num = num // 10
    return origianl == result

print(check_palindrome(2323))
print(check_palindrome(2567))
print(check_palindrome(2222))
```

```python
def is_palindrome(num):
    return str(num) == str(num)[::-1]

print(is_palindrome(455))
print(is_palindrome(4554))
```

```python
def is_palindrome_new(num):
    s = str(num)
    reverse = ''.join(reversed(s))
    return s == reverse

print(is_palindrome_new(2323))
print(is_palindrome_new(1212))
print(is_palindrome_new(121))
```

```python
def is_palindrome2(num):
    s = str(num)
    left = 0
    right = len(s) -1

    while left < right:
        if s[left] != s[right]:
            return False

        left +=1
        right -=1
    return True
print(is_palindrome2(4554))
print(is_palindrome2(89))
print(is_palindrome2(453))
```

```python
def palindrome_recursion(s):
    if len(s) <= 1:
        return True

    if s[0] != s[-1]:
        return False

    return palindrome_recursion(s[1:-1])

print(palindrome_recursion("madam"))
print(palindrome_recursion("mahesh"))
```

```python
def is_palindrome_optimized(num):
    if num < 0 or (num %10 ==0 and num != 0):
        return False
    reverse = 0
    while num > reverse:
        reverse = reverse * 10 + num % 10
        num //= 10
    return num == reverse or reverse// 10

print(is_palindrome_optimized(2323))
print(is_palindrome_optimized(2332))
```

```python
def armstrong_number(num):
    total = 0
    nod = len(str(num))

    while num > 0:
        id = num % 10
        total = total + (id ** nod)
        num //= 10
    return total == num

print(armstrong_number(3434))
print(armstrong_number(3443))
```
