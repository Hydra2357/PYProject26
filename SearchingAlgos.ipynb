Binary Search,O(log n),O(1),Yes,Large sorted arrays\
Jump Search,O(√n),O(1),Yes,Medium-large sorted arrays\

Ternary Search,O(log n),O(1),Yes,Unimodal functions\

```python
def linear_search(arr, key):
    for i in range(len(arr)):
        if arr[i] == key:
            return i
    return -1
```

```python
def binary_search(arr, key):
    low, high = 0, len(arr) - 1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == key:
            return mid
        elif arr[mid] < key:
            low = mid + 1
        else:
            high = mid - 1
    return -1
```
