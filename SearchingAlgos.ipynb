```python
| Algorithm            | Data Requirement | Time Complexity  |
| -------------------- | ---------------- | ---------------- |
| Linear Search        | Unsorted         | O(n)             |
| Binary Search        | Sorted           | O(log n)         |
| Jump Search          | Sorted           | O(√n)            |
| Interpolation Search | Uniform Sorted   | O(log log n) avg |
| Exponential Search   | Sorted           | O(log n)         |
```

```python
def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1


def binary_search(arr, target):
    left = 0
    right = len(arr) - 1

    while left <= right:
        mid = (left + right) // 2

        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return -1


# Main program
if __name__ == "__main__":
    arr = list(map(int, input("Enter numbers separated by space: ").split()))
    target = int(input("Enter target number: "))

    print("Linear Search Index:", linear_search(arr, target))

    # Binary search works only on sorted array
    arr.sort()
    print("Sorted array for binary search:", arr)
    print("Binary Search Index:", binary_search(arr, target))
```

Jump Search — O(√n)
Interpolation Search — O(log log n) (uniform data)
Exponential Search — O(log n)
Ternary Search — O(log₃n)

```python
import math


# 1. Linear Search
def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1


# 2. Binary Search
def binary_search(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1

    return -1


# 3. Jump Search
def jump_search(arr, target):
    n = len(arr)
    step = int(math.sqrt(n))
    prev = 0

    while prev < n and arr[min(step, n) - 1] < target:
        prev = step
        step += int(math.sqrt(n))

        if prev >= n:
            return -1

    for i in range(prev, min(step, n)):
        if arr[i] == target:
            return i

    return -1


# 4. Interpolation Search
def interpolation_search(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high and arr[low] <= target <= arr[high]:

        if low == high:
            if arr[low] == target:
                return low
            return -1

        pos = low + (
            (target - arr[low])
            * (high - low)
            // (arr[high] - arr[low])
        )

        if arr[pos] == target:
            return pos

        elif arr[pos] < target:
            low = pos + 1

        else:
            high = pos - 1

    return -1


# 5. Exponential Search
def exponential_search(arr, target):

    if arr[0] == target:
        return 0

    i = 1

    while i < len(arr) and arr[i] <= target:
        i *= 2

    low = i // 2
    high = min(i, len(arr) - 1)

    while low <= high:

        mid = (low + high) // 2

        if arr[mid] == target:
            return mid

        elif arr[mid] < target:
            low = mid + 1

        else:
            high = mid - 1

    return -1


# 6. Ternary Search
def ternary_search(arr, target):
    left = 0
    right = len(arr) - 1

    while left <= right:

        mid1 = left + (right - left) // 3
        mid2 = right - (right - left) // 3

        if arr[mid1] == target:
            return mid1

        if arr[mid2] == target:
            return mid2

        if target < arr[mid1]:
            right = mid1 - 1

        elif target > arr[mid2]:
            left = mid2 + 1

        else:
            left = mid1 + 1
            right = mid2 - 1

    return -1


# DRIVER CODE
arr = [2, 5, 8, 12, 16, 22, 30, 45, 60]
target = 22

print("Linear:", linear_search(arr, target))
print("Binary:", binary_search(arr, target))
print("Jump:", jump_search(arr, target))
print("Interpolation:", interpolation_search(arr, target))
print("Exponential:", exponential_search(arr, target))
print("Ternary:", ternary_search(arr, target))
```
