```python
# searching_algorithms.py

# 1. Linear Search
def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1


# 2. Binary Search (Iterative)
def binary_search(arr, target):
    left, right = 0, len(arr) - 1

    while left <= right:
        mid = (left + right) // 2

        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return -1


# 3. Binary Search (Recursive)
def binary_recursive(arr, left, right, target):
    if left > right:
        return -1

    mid = (left + right) // 2

    if arr[mid] == target:
        return mid
    elif arr[mid] < target:
        return binary_recursive(arr, mid + 1, right, target)

    return binary_recursive(arr, left, mid - 1, target)


# 4. Jump Search
import math

def jump_search(arr, target):
    n = len(arr)
    step = int(math.sqrt(n))
    prev = 0

    while prev < n and arr[min(step, n)-1] < target:
        prev = step
        step += int(math.sqrt(n))

    for i in range(prev, min(step, n)):
        if arr[i] == target:
            return i

    return -1


# 5. Interpolation Search
def interpolation_search(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high and low < len(arr) and \
          arr[low] <= target <= arr[high]:

        if arr[low] == arr[high]:
            if arr[low] == target:
                return low
            return -1

        pos = low + (
            (target - arr[low]) *
            (high - low)
        ) // (arr[high] - arr[low])

        if arr[pos] == target:
            return pos

        if arr[pos] < target:
            low = pos + 1
        else:
            high = pos - 1

    return -1


# 6. Exponential Search
def exponential_search(arr, target):
    n = len(arr)

    if arr[0] == target:
        return 0

    i = 1

    while i < n and arr[i] <= target:
        i *= 2

    left = i // 2
    right = min(i, n - 1)

    return binary_recursive(arr, left, right, target)


# Example
arr = [2, 4, 7, 10, 13, 18, 22, 29]

print("Linear:", linear_search(arr, 13))
print("Binary:", binary_search(arr, 13))
print("Binary Recursive:", binary_recursive(arr, 0, len(arr)-1, 13))
print("Jump:", jump_search(arr, 13))
print("Interpolation:", interpolation_search(arr, 13))
print("Exponential:", exponential_search(arr, 13))
```
