```python
class Book:
    def __init__(self, title, author, year, isbn=None):
        self.title = title
        self.author = author
        self.year = year
        self.isbn = isbn

    def display(self):
        print(f"Title     : {self.title}")
        print(f"Author    : {self.author}")
        print(f"Year      : {self.year}")
        if self.isbn:
            print(f"ISBN      : {self.isbn}")
        print("-" * 40)

    # ==================== Alternative Constructors ====================

    @classmethod
    def from_string(cls, string):
        """Create Book from comma-separated string: 'title,author,year,isbn'"""
        try:
            parts = [part.strip() for part in string.split(",")]
            if len(parts) == 3:
                title, author, year = parts
                return cls(title, author, int(year))
            elif len(parts) == 4:
                title, author, year, isbn = parts
                return cls(title, author, int(year), isbn)
            else:
                raise ValueError("Invalid format")
        except Exception:
            raise ValueError(f"Invalid string format. Use: 'title,author,year' or 'title,author,year,isbn'")

    @classmethod
    def from_dict(cls, data):
        """Create Book from dictionary"""
        return cls(
            title=data.get('title'),
            author=data.get('author'),
            year=data.get('year'),
            isbn=data.get('isbn')
        )

    @classmethod
    def from_json(cls, json_string):
        """Create Book from JSON string"""
        import json
        data = json.loads(json_string)
        return cls.from_dict(data)   # Reusing from_dict for clean code


# ====================== Usage Examples ======================

print("1. Creating using normal constructor:")
book1 = Book("The Python Book", "Mahesh Sharma", 2025)
book1.display()

print("2. Creating using from_string (Class Method):")
book2 = Book.from_string("Atomic Habits, James Clear, 2018, 9780735211292")
book2.display()

print("3. Creating using from_string on an existing instance:")
book3 = book2.from_string("Deep Work, Cal Newport, 2016")
book3.display()

print("4. Creating using from_dict:")
book4 = Book.from_dict({
    'title': "Clean Code",
    'author': "Robert C. Martin",
    'year': 2008,
    'isbn': "9780132350884"
})
book4.display()

print("5. Creating using from_json:")
json_data = '{"title": "The Pragmatic Programmer", "author": "Andrew Hunt", "year": 1999}'
book5 = Book.from_json(json_data)
book5.display()
```

```python
# Super() with multiple inheritance
# MRO - method resolution order
class A:
    def __init__(self):
        print("A __init__ called ...")
        super().__init__()

class B:
    def __init__(self):
        print("B __init__ called ...")
        super().__init__()

class C(A, B):
    def __init__(self):
        super().__init__()
        print("C __init__ called ...")

c = C()
print(C.__mro__)
```

```python
class Logger:
    def log(self, msg):
        print(f"LOG: {msg}")
        super().log(msg) if hasattr(super(), 'log') else None

class TimestampLogger(Logger):
    def log(self, msg):
        import datetime
        ts = datetime.datetime.now().strftime("%H:%M:%S")
        super().log(f"[{ts}] {msg}")

class FileLogger(Logger):
    def log(self, msg):
        print(f"FILE: {msg}")
        super().log(msg)

class AdvanceLogger(TimestampLogger, FileLogger):
    def log(self, msg):
        print("Advanced LOgger start")
        super().log(msg)

adv = AdvanceLogger()
adv.log("system started...")
```

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

    def perimeter(self):
        return 2 * 3.14 * self.radius

# Usage
rect = Rectangle(10, 5)
print("Rectangle Area:", rect.area())        # 50
print("Rectangle Perimeter:", rect.perimeter())

circ = Circle(7)
print("Circle Area:", circ.area())
```

```python
# Mixin 1: Adds logging capability
class LogMixin:
    def log(self, message):
        print(f"[LOG] {self.__class__.__name__}: {message}")

# Mixin 2: Adds serialization
class JSONSerializableMixin:
    def to_json(self):
        import json
        return json.dumps(vars(self))   # uses attributes from the host class

# Main class (can be used with or without mixins)
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

# Using mixins with multiple inheritance
class Employee(LogMixin, JSONSerializableMixin, Person):
    def __init__(self, name, age, employee_id):
        super().__init__(name, age)      # calls Person.__init__
        self.employee_id = employee_id

    def work(self):
        print(f"{self.name} is working")
        self.log("Started working")      # from LogMixin

# Usage
emp = Employee("Mahesh", 30, "E123")
emp.work()
print(emp.to_json())                 # from JSONSerializableMixin
```

```python
class LoggerMixin:
    def log(self, msg):
        print(f"LOG: {msg}")

class TimerMixin:
    def show_time(self):
        from datetime import datetime
        print("Current time:", datetime.now().strftime("%H:%M:%S"))

class Employee:
    def __init__(self, name):
        self.name = name

    def work(self):
        print(f"{self.name} is working...")

class Manager(LoggerMixin, TimerMixin, Employee):
    def manage(self):
        self.log("Manager started managing")
        self.show_time()
        self.work()

m1 = Manager("Hydra")
m1.manage()
```

```python
class SaveMixin:
    def save(self):
        print(f"Saving {self.name} to database...")

class DeleteMixin:
    def delete(self):
        print(f"Deleting {self.name}...")

class User:
    def __init__(self, name):
        self.name = name

class AdminUser(SaveMixin, DeleteMixin, User):
    def login(self):
        print(f"{self.name} logged in as Admin")

admin = AdminUser("Mahesh")
admin.login()
admin.save()
admin.delete()
```

```python
class Employee:
    def __init__(self, name):
        self.name = name

    def show(self):
        print(f"{self.name} is from class Employee")

class Dancer:
    def __init__(self, dance):
        self.dance = dance

    def show(self):
        print(f"{self.dance} is from class Dance")

class DancerEmployee(Employee, Dancer):
    def __init__(self, dance, name):
        super().__init__(name)
        self.dance = dance
        self.name = name

o1 = DancerEmployee("kathak", "shivaji")
print(o1.name)
print(o1.dance)
print(o1.show())
print(DancerEmployee.mro())
```

```python
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

    def start(self):
        return f"This Vehicle having {self.brand}."

class Car(Vehicle):
    def __init__(self, brand, model, cost):
        super().__init__(brand)
        self.cost = cost
        self.model = model
        print(f"Car {model} created...")

    def start(self):
        super().start()
        print(f"THe Car has model {self.model} and cost {self.cost}")
        return f"{self.model} car engine roaring!"

class Electric_car(Car):
    def __init__(self, brand, model, cost, battery_size):
        super().__init__(brand, model, cost)
        self.battery_size = battery_size
        print(f"Tesla : {model}")
        print(f"Electric car have current battery : {battery_size} %")

    def start(self):
        super().start()
        return f"This is start method from electric car with model {self.model} and {self.battery_size} %"

e1 = Electric_car("Tesla", "Model X (luxury SUV)", "2cr", 89)
print(e1.start())
```

```python
class A:
    def method(self):
        print("this is from class A")

class B(A):
    def method(self):
        print("This is from class B")

class C(A):
    def method(self):
        print("This is from the class C")

class D(B, C):
    pass

d = D()
d.method()
print(D.__mro__)
print()
print("MRO for D: ", [cls.__name__ for cls in D.__mro__])
```

```python
class Character:
    """Base class for all game characters"""
    def __init__(self, name, level=1):
        self.name = name
        self.level = level
        self.health = 100 + (level * 20)
        self.mana = 50 + (level * 10)
        print(f"[{self.name}] Base Character created (Level {level})")

    def attack(self):
        damage = 10 + (self.level * 2)
        print(f"[{self.name}] Basic attack! Dealt {damage} damage.")
        return damage

    def take_damage(self, damage):
        self.health -= damage
        print(f"[{self.name}] Took {damage} damage. Health left: {self.health}")
        if self.health <= 0:
            print(f"[{self.name}] has been defeated!")

    def show_status(self):
        print(f"\n=== Status of {self.name} ===")
        print(f"Level: {self.level} | Health: {self.health} | Mana: {self.mana}")


class Warrior(Character):
    """Warrior specialization"""
    def __init__(self, name, level=1, strength=15):
        super().__init__(name, level)
        self.strength = strength
        self.shield = 30
        print(f"[{self.name}] Warrior initialized with Strength: {strength}")

    def attack(self):
        damage = 25 + (self.strength * 1.5) + (self.level * 3)
        print(f"[{self.name}] Warrior swings sword! Dealt {damage:.0f} physical damage.")
        return damage

    def use_shield(self):
        print(f"[{self.name}] Raises shield! Defense increased.")
        self.health += 10  # Small heal from shield


class Mage(Character):
    """Mage specialization"""
    def __init__(self, name, level=1, intelligence=18):
        super().__init__(name, level)
        self.intelligence = intelligence
        self.spell_power = 20
        print(f"[{self.name}] Mage initialized with Intelligence: {intelligence}")

    def attack(self):
        if self.mana >= 15:
            damage = 30 + (self.intelligence * 1.8) + (self.level * 4)
            self.mana -= 15
            print(f"[{self.name}] Mage casts Fireball! Dealt {damage:.0f} magic damage. Mana left: {self.mana}")
            return damage
        else:
            print(f"[{self.name}] Not enough mana for spell!")
            return super().attack()

    def heal(self, amount=25):
        if self.mana >= 20:
            self.health += amount
            self.mana -= 20
            print(f"[{self.name}] Mage casts Heal! Restored {amount} health. Mana left: {self.mana}")
        else:
            print(f"[{self.name}] Not enough mana to heal!")


class Paladin(Warrior, Mage):
    """Hybrid Class: Warrior + Mage = Paladin (Holy Knight)"""
    def __init__(self, name, level=1):
        # Python handles the diamond automatically via MRO
        super().__init__(name, level)   # Calls Warrior.__init__ first (due to MRO)
        self.holy_power = 25
        print(f"[{self.name}] Paladin (Holy Warrior) fully awakened!")

    def attack(self):
        print(f"[{self.name}] Paladin combines holy power with sword and magic!")
        # Call Warrior's attack (physical)
        physical_dmg = Warrior.attack(self)
        # Call Mage's attack (magical) - but safely using super() chain
        if self.mana >= 10:
            magical_dmg = Mage.attack(self)
            total_damage = physical_dmg + magical_dmg
            print(f"[{self.name}] Holy Strike! Total Damage: {total_damage:.0f}")
            return total_damage
        return physical_dmg

    def holy_smite(self):
        if self.mana >= 30:
            damage = 45 + (self.level * 5) + self.holy_power
            self.mana -= 30
            print(f"[{self.name}] Casts Holy Smite! Dealt {damage} radiant damage!")
            return damage
        else:
            print(f"[{self.name}] Not enough mana for Holy Smite!")
            return self.attack()

    def show_status(self):
        super().show_status()   # This will follow MRO
        print(f"Holy Power: {self.holy_power} | Class: Paladin")

if __name__ == "__main__":
    print("=== Fantasy RPG - Hybrid Inheritance Demo ===\n")

    # Create a Paladin (triggers the diamond problem)
    arthur = Paladin("Arthur the Holy Knight", level=5)

    print("\n--- Battle Simulation ---")
    arthur.show_status()

    arthur.attack()
    arthur.holy_smite()
    arthur.use_shield()        # From Warrior
    arthur.heal(40)            # From Mage

    print("\n--- Final Status ---")
    arthur.show_status()

    # Show how Python resolved the Diamond Problem
    print("\nMethod Resolution Order (MRO) for Paladin:")
    print([cls.__name__ for cls in Paladin.__mro__])
```
