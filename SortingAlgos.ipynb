```python
def bubble_sort_ultra_optimized(arr):
    n = len(arr)
    last_swap = n - 1  # initially assume we need to check till end

    for i in range(n):
        swapped = False
        current_last_swap = -1

        for j in range(0, last_swap):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                swapped = True
                current_last_swap = j  # remember last position where swap happened

        if not swapped:
            break

        last_swap = current_last_swap  # next pass doesn't need to go beyond last swap

    return arr

arr = [5, 1, 4, 2, 8, 0, 2]
print(bubble_sort_ultra_optimized(arr.copy()))
```

```python
def bubble_sort_basics(arr):
    n = len(arr)

    for i in range(n):
        for j in range(0, n-i-1):

            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    return arr

arr = [23 ,34,45 ,56,67,78,56,78,21]
print("Original arr: ", arr)
bubble_sort_basics(arr)
print("Sorted arr: ", arr)
```

```python
def optimized_bubble_sort(arr):
    n = len(arr)
    swapped = True

    for i in range(n):
        if not swapped:
            break
        swapped = False

        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                swapped = True
        return arr

if __name__ == "__main__":
    numbers = [23 ,34 ,45, 56 ,67,78,89,212]
    print("Original Array: ", numbers)
    optimized_bubble_sort(numbers)
    print("Sorted Array: ", numbers)
```

```python
def optimized_bubble_sort_with_steps(arr):
    n = len(arr)
    print(f"Original array: {arr}\n")

    for i in range(n):
        swapped = False
        print(f"Pass {i+1}:")

        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
                print(f"  Swapped {arr[j+1]} ↔ {arr[j]}  →  {arr}")
            else:
                print(f"  No swap at index {j}")

        print(f"  Swapped in this pass: {swapped}\n")
        if not swapped:
            print("✅ Array is already sorted! Stopping early.")
            break
    return arr

arr = [64, 34, 25, 12, 22, 11, 90]
optimized_bubble_sort_with_steps(arr)
print("Final Sorted Array:", arr)
```

```python
#Recursive Bubble sort
def bubble_sort_recursive(arr, n=None):
    if n is None:
        n = len(arr)

    if n==1:
        return arr

    for i in range(n-1):
        if arr[i] > arr[i+1]:
            arr[i], arr[i+1] = arr[i+1], arr[i]

    bubble_sort_recursive(arr, n-1)
    return arr

arr = [64, 34, 25, 12, 22, 11, 90]
bubble_sort_recursive(arr)
print("Final Sorted Array:", arr)
```

```python
def bubble_sort_descending(arr):
    n = len(arr)
    for i in range(n):
        swapp = False
        for j in range(0, n-i-1):
            if arr[j] < arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                swapp = True
        if not swapp:
            break
    return arr

arr = [364, 34, 325, 12, 222, 11, 90]
bubble_sort_descending(arr)
print("Array sorting in descending order:", arr)
```

```python
def bubble_sort(arr, reverse=True):
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(0, n-i-1):

            if(not reverse and arr[j] > arr[j+1]) or (reverse and arr[j] < arr[j+1]):
                arr[j], arr[j+1] = arr[j+1], arr[j]
                swapped = True
        if not swapped:
            break
    return arr

arr = [364, 34, 325, 12, 222, 11, 90]
bubble_sort(arr, reverse=True)
print("Array sorting in ascending order:", arr)
```

```python
def insertion_sort(arr):
    n = len(arr)

    for i in range(1, n):
        key = arr[i]
        j = i-1

        while j>=0 and arr[j] > key:
            arr[j+1] = arr[j]
            j -= 1

        arr[j+1] = key
    return arr

arr = [364, 34, 325, 12, 222, 11, 90]
insertion_sort(arr)
print("Array sorting in ascending order using Insertion sort:\n", arr)
```
