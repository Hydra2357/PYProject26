```python
def bubble_sort_ultra_optimized(arr):
    n = len(arr)
    last_swap = n - 1  # initially assume we need to check till end

    for i in range(n):
        swapped = False
        current_last_swap = -1

        for j in range(0, last_swap):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                swapped = True
                current_last_swap = j  # remember last position where swap happened

        if not swapped:
            break

        last_swap = current_last_swap  # next pass doesn't need to go beyond last swap

    return arr

arr = [5, 1, 4, 2, 8, 0, 2]
print(bubble_sort_ultra_optimized(arr.copy()))
```

```python
def optimized_bubble_sort(arr):
    n = len(arr)
    swapped = True

    for i in range(n):
        if not swapped:
            break
        swapped = False

        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                swapped = True
        return arr

if __name__ == "__main__":
    numbers = [23 ,34 ,45, 56 ,67,78,89,212]
    print("Original Array: ", numbers)
    optimized_bubble_sort(numbers)
    print("Sorted Array: ", numbers)
```

```python
def optimized_bubble_sort_with_steps(arr):
    n = len(arr)
    print(f"Original array: {arr}\n")

    for i in range(n):
        swapped = False
        print(f"Pass {i+1}:")

        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
                print(f"  Swapped {arr[j+1]} ↔ {arr[j]}  →  {arr}")
            else:
                print(f"  No swap at index {j}")

        print(f"  Swapped in this pass: {swapped}\n")
        if not swapped:
            print("✅ Array is already sorted! Stopping early.")
            break
    return arr

arr = [64, 34, 25, 12, 22, 11, 90]
optimized_bubble_sort_with_steps(arr)
print("Final Sorted Array:", arr)
```

```python
#Recursive Bubble sort
def bubble_sort_recursive(arr, n=None):
    if n is None:
        n = len(arr)

    if n==1:
        return arr

    for i in range(n-1):
        if arr[i] > arr[i+1]:
            arr[i], arr[i+1] = arr[i+1], arr[i]

    bubble_sort_recursive(arr, n-1)
    return arr

arr = [64, 34, 25, 12, 22, 11, 90]
bubble_sort_recursive(arr)
print("Final Sorted Array:", arr)
```

```python
def bubble_sort(arr, reverse=True):
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(0, n-i-1):

            if(not reverse and arr[j] > arr[j+1]) or (reverse and arr[j] < arr[j+1]):
                arr[j], arr[j+1] = arr[j+1], arr[j]
                swapped = True
        if not swapped:
            break
    return arr

arr = [364, 34, 325, 12, 222, 11, 90]
bubble_sort(arr, reverse=True)
print("Array sorting in ascending order:", arr)
```

```python
def insertion_sort(arr):
    n = len(arr)

    for i in range(1, n):
        key = arr[i]
        j = i-1

        while j>=0 and arr[j] > key:
            arr[j+1] = arr[j]
            j -= 1

        arr[j+1] = key
    return arr

arr = [364, 34, 325, 12, 222, 11, 90]
insertion_sort(arr)
print("Array sorting in ascending order using Insertion sort:\n", arr)
```

```python
def bubble_sort_chai(a):
    """Bubble sort using chai or code"""
    n = len(a)
    for i in range(n):
        for j in range(0, n-i-1):
            if a[j] > a[j+1]:
                a[j], a[j+1] = a[j+1], a[j]
    return a

a = [121, 23 , 122, 56, 78, 89]
print(a)
bubble_sort_chai(a)
print(a)
```

```python
def selection_sort_chai(a):
    """Selection sort using chai or code"""
    n = len(a)
    for i  in range(n):
        min = i
        for j in range(i, n):
            if a[min] > a[j]:
                min = j
        a[i], a[min] = a[min], a[i]

a = [121, 23 , 122, 56, 78, 89]
print(a)
selection_sort_chai(a)
print(a)
```

```python
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        # Find the minimum element in remaining unsorted array
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j

        # Swap the found minimum element with the first element
        arr[i], arr[min_idx] = arr[min_idx], arr[i]

    return arr


# Example usage
arr = [64, 25, 12, 22, 11]
print(selection_sort(arr))  # Output: [11, 12, 22, 25, 64]
```

```python
def selection_sort_descending(arr):
    n = len(arr)
    for i in range(n):
        # Find the maximum element in remaining unsorted array
        max_idx = i
        for j in range(i + 1, n):
            if arr[j] > arr[max_idx]:
                max_idx = j

        # Swap
        arr[i], arr[max_idx] = arr[max_idx], arr[i]

    return arr


arr = [64, 25, 12, 22, 11]
print(selection_sort_descending(arr))  # Output: [64, 25, 22, 12, 11]
```

```python
def selection_sort_optimized(arr):
    n = len(arr)
    for i in range(n - 1):  # No need for last iteration
        min_idx = i
        swapped = False

        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
                swapped = True

        arr[i], arr[min_idx] = arr[min_idx], arr[i]

        # If no swap needed in this pass, array is sorted
        if not swapped and i == 0:  # Only useful in best case
            break

    return arr
```

```python
def selection_sort_recursive(arr, start=0):
    n = len(arr)

    if start >= n - 1:
        return arr

    # Find minimum in remaining array
    min_idx = start
    for i in range(start + 1, n):
        if arr[i] < arr[min_idx]:
            min_idx = i

    # Swap
    arr[start], arr[min_idx] = arr[min_idx], arr[start]

    # Recursive call
    return selection_sort_recursive(arr, start + 1)


arr = [64, 25, 12, 22, 11]
print(selection_sort_recursive(arr))
```

```python
def selection_sort_key(arr, key=lambda x: x):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if key(arr[j]) < key(arr[min_idx]):
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr


# Examples:
data = ["apple", "banana", "kiwi", "mango"]
print(selection_sort_key(data, key=len))           # Sort by length

students = [
    {"name": "Alice", "age": 25},
    {"name": "Bob", "age": 20},
    {"name": "Charlie", "age": 23}
]
print(selection_sort_key(students, key=lambda x: x["age"]))
"""Generic Selection Sort with Key Function"""
```

```python
def selection_sort_new(arr):
    arr = arr.copy()  # Don't modify original
    n = len(arr)
    sorted_arr = []

    while arr:
        min_val = min(arr)
        sorted_arr.append(min_val)
        arr.remove(min_val)  # O(n) operation - makes it less efficient

    return sorted_arr
"""Selection Sort Returning New List"""
```
