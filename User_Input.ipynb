```python
def remainder_without_mod(n: int, x: int) -> int:
    if x == 0:
        raise ValueError("Divisor cannot be zero")
    # integer quotient
    q = n // x
    # remainder formula
    r = n - x * q
    return r


def main():
    try:
        n = int(input("Enter dividend (n): ").strip())
        x = int(input("Enter divisor (x): ").strip())

        r = remainder_without_mod(n, x)

        print("\nResults")
        print("Dividend:", n)
        print("Divisor:", x)
        print("Quotient:", n // x)
        print("Remainder:", r)

    except ValueError as e:
        print("Error:", e)


if __name__ == "__main__":
    main()
```

```python
# binary search method to find the square root of a number. The idea is to search the value whose square is closest to n.

def sqr_root(n45):
    l = 0
    ans = 0
    r = n45

    while l <= r:
        mid = l + (r - l) // 2
        if mid * mid == n45:
            return mid

        if mid * mid > n45:
            r = mid - 1
        else:
            ans = mid
            l = mid + 1

    return ans


n45 = int(input("Enter the no for root:"))
print(f"for {n45} :", sqr_root(n45))
print(sqr_root(23))
print(sqr_root(99))
print(sqr_root(9))
```

```python
def traillingZeros(n):
    '''TrailingZeros(n!)
= ⌊n/5⌋ + ⌊n/25⌋ + ⌊n/125⌋ + ⌊n/625⌋ + ...'''
    count = 0

    while n > 0:
        n //= 5
        count += n
    return count


n = int(input())
print(traillingZeros(n))
# print(traillingZeros(__doc__))
```

```python
def get_xor_upto(n):
    if n < 0: return 0
    remainder1 = n % 4
    if remainder1 == 0: return n
    if remainder1 == 1: return 1
    if remainder1 == 2: return n + 1
    return 0


def find_missing_number_adv(arr, n):
    theoretical_xor = get_xor_upto(n)

    actual_xor = 0
    for num34 in arr:
        actual_xor ^= num34
    return theoretical_xor ^ actual_xor


if __name__ == "__main__":
    ex1_n = 5
    ex1_arr = [2, 3, 1, 5]
    print(f"Input: N={ex1_n}, arr={ex1_arr}")
    print(f"Missing Number: {find_missing_number_adv(ex1_arr, ex1_n)}")

    print("-" * 30)
    ex2_n = 6
    ex2_arr = [4, 5, 1, 2, 3]
    print(f"Input: N={ex2_n}, arr={ex2_arr}")
    print(f"Missing Number: {find_missing_number_adv(ex2_arr, ex2_n)}")
```

```python
import sys


# -------- Iterative Fibonacci --------
def fibonacci_iterative(n):
    a, b = 0, 1
    series = []

    for _ in range(n):
        series.append(a)
        a, b = b, a + b

    return series


# -------- Recursive Fibonacci with Memoization --------
def fibonacci_recursive(n, memo={}):
    if n in memo:
        return memo[n]

    if n <= 1:
        return n

    memo[n] = fibonacci_recursive(n - 1, memo) + fibonacci_recursive(n - 2, memo)
    return memo[n]


def fibonacci_recursive_series(n):
    return [fibonacci_recursive(i) for i in range(n)]


# -------- Fast Doubling Method (Very Efficient O(log n)) --------
def fib_fast_doubling(n):
    if n == 0:
        return (0, 1)

    a, b = fib_fast_doubling(n // 2)
    c = a * (2 * b - a)
    d = a * a + b * b

    if n % 2 == 0:
        return (c, d)
    else:
        return (d, c + d)


def fibonacci_fast_series(n):
    return [fib_fast_doubling(i)[0] for i in range(n)]


# -------- Display Function --------
def display(series):
    print("\nFibonacci Series:")
    for num in series:
        print(num, end=" ")
    print()


# -------- Main Program --------
def main():
    try:
        n = int(input("Enter number of terms: "))

        if n <= 0:
            print("Please enter a positive integer.")
            sys.exit()

        print("\nChoose Method:")
        print("1. Iterative")
        print("2. Recursive with Memoization")
        print("3. Fast Doubling (Advanced)")

        choice = int(input("Enter choice (1-3): "))

        if choice == 1:
            series = fibonacci_iterative(n)

        elif choice == 2:
            series = fibonacci_recursive_series(n)

        elif choice == 3:
            series = fibonacci_fast_series(n)

        else:
            print("Invalid choice.")
            sys.exit()

        display(series)

    except ValueError:
        print("Invalid input. Please enter integers only.")


# -------- Program Entry Point --------
if __name__ == "__main__":
    main()
```

```python
def max_occurring_char(s):
    freq = {}

    for ch in s:
        if ch in freq:
            freq[ch] += 1
        else:
            freq[ch] = 1

    max_char = None
    max_count = 0

    for ch in freq:
        if freq[ch] > max_count:
            max_count = freq[ch]
            max_char = ch

    return max_char, max_count


string = input("Enter the string: ")
char, count = max_occurring_char(string)
print(f"The given string is {string}")
print(f"The given char is {char}")

print("Max occurring character: ", char)
print("frequency: ", count)
```

```python
#expception groups
def process_tasks():
    exceptions = []
    for i in range(3):
        try:
            if i == 1:
                raise ValueError("Task 1 failed")
            if i == 2:
                raise ValueError("Task 2 failed")
        except Exception as rt:
            exceptions.append(rt)

    if exceptions:
        raise ExceptionGroup("Multiple Tasks failed..", exceptions)


try:
    process_tasks()
except* ValueError as eg:
    print("ValueErrors: ", eg.exceptions)
except* TypeError as eg:
    print("TypeErrors: ", eg.exceptions)
```

```python
# Define your own exception
class InsufficientBalanceError(Exception):
    def __init__(self, balance, amount):
        self.balance = balance
        self.amount = amount
        super().__init__(f"Insufficient balance! Available: ₹{balance}, Tried to withdraw: ₹{amount}")


class InvalidEmailError(Exception):
    """Custom exception for invalid email"""
    pass


# Using custom exceptions
def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientBalanceError(balance, amount)
    return balance - amount


def send_email(email):
    if "@" not in email or "." not in email:
        raise InvalidEmailError(f"Invalid email format: {email}")


# Test
try:
    withdraw(5000, 10000)
except InsufficientBalanceError as e:
    print(e)

try:
    send_email("maheshgmailcom")
except InvalidEmailError as e:
    print(e)
```

```python
import random


def check_winner(user, computer):
    if user == computer:
        return "Draw"

    win_map = {
        's': 'w',  # Snake beats Water
        'w': 'g',  # Water beats Gun
        'g': 's'  # Gun beats Snake
    }

    if win_map[user] == computer:
        return "Win"
    else:
        return "Lose"


choices = ['s', 'w', 'g']

user = input("Enter (s/w/g): ").lower()

if user not in choices:
    print("Invalid input")
else:
    computer = random.choice(choices)

    print(f"You: {user}, Computer: {computer}")
    print("Result:", check_winner(user, computer))
```

```python
import random

choices = ['s', 'w', 'g']
score_user = 0
score_comp = 0


def check(user, comp):
    if user == comp:
        return 0
    elif (user == 's' and comp == 'w') or \
            (user == 'w' and comp == 'g') or \
            (user == 'g' and comp == 's'):
        return 1
    else:
        return -1


rounds = 5

for i in range(rounds):
    print(f"\nRound {i + 1}")
    user = input("Enter (s/w/g): ").lower()

    if user not in choices:
        print("Invalid input")
        continue

    comp = random.choice(choices)

    print("Computer:", comp)

    result = check(user, comp)

    if result == 1:
        print("You Win this round")
        score_user += 1
    elif result == -1:
        print("Computer Wins this round")
        score_comp += 1
    else:
        print("Draw")

print("\nFinal Score")
print("You:", score_user)
print("Computer:", score_comp)

if score_user > score_comp:
    print("Overall Winner: You")
elif score_user < score_comp:
    print("Overall Winner: Computer")
else:
    print("Match Draw")
```

```python
import random


def snake_water_gun_game():
    print("Welcome to Snake Water Gun Game! 🐍 💧 🔫")
    print("Rules:")
    print("🐍 Snake drinks Water 💧 → Snake wins")
    print("💧 Water puts out Gun 🔫 → Water wins")
    print("🔫 Gun shoots Snake 🐍 → Gun wins")
    print("Same choices = Draw\n")

    choices = ['snake', 'water', 'gun']
    emojis = {'snake': '🐍', 'water': '💧', 'gun': '🔫'}

    user_score = 0
    computer_score = 0
    rounds = 0

    while True:
        print(f"\n--- Round {rounds + 1} ---")
        print("Your Score:", user_score, "| Computer Score:", computer_score)

        # User input
        user_choice = input("\nEnter your choice (snake/water/gun) or 'quit' to exit: ").lower().strip()

        if user_choice == 'quit':
            break

        if user_choice not in choices:
            print("Invalid choice! Please choose snake, water, or gun.")
            continue

        # Computer choice
        computer_choice = random.choice(choices)

        print(f"You chose: {user_choice.capitalize()} {emojis[user_choice]}")
        print(f"Computer chose: {computer_choice.capitalize()} {emojis[computer_choice]}")

        # Determine winner
        if user_choice == computer_choice:
            print("It's a Draw! 🤝")
        elif (user_choice == 'snake' and computer_choice == 'water') or \
                (user_choice == 'water' and computer_choice == 'gun') or \
                (user_choice == 'gun' and computer_choice == 'snake'):
            print("You Win! 🎉")
            user_score += 1
        else:
            print("Computer Wins! 💻")
            computer_score += 1

        rounds += 1

        # Ask to play again
        play_again = input("\nDo you want to play another round? (y/n): ").lower().strip()
        if play_again != 'y':
            break

    # Final result
    print("\n" + "=" * 40)
    print("Game Over! Final Score:")
    print(f"You: {user_score} | Computer: {computer_score}")

    if user_score > computer_score:
        print("Congratulations! You are the overall winner! 🏆")
    elif computer_score > user_score:
        print("Computer is the overall winner! Better luck next time!")
    else:
        print("It's a tie overall! 🤝")

    print("=" * 40)


# Run the game
if __name__ == "__main__":
    snake_water_gun_game()
```

```python
import time
from typing import Optional, Final


def clear_screen():
    print("\033[H\033[J", end="")  # ANSI clear


class Game:
    def __init__(self):
        self.running = True
        self.score = 0
        self.player_hp = 100
        self.frame_count = 0

    def handle_input(self) -> None:
        # Non-blocking input is more complex → here we use simple blocking
        action = input("→ ").lower().strip()
        if action in ("q", "quit", "exit"):
            self.running = False
        elif action == "attack":
            self.score += 15
            print("→ Dealt damage!")
        elif action == "heal":
            self.player_hp = min(100, self.player_hp + 30)
            print("→ HP restored")

    def update(self) -> None:
        self.frame_count += 1
        # Simulate enemy attack every ~4 seconds
        if self.frame_count % 40 == 0:
            damage = 12
            self.player_hp -= damage
            print(f"Enemy attacked! -{damage} HP")

    def render(self) -> None:
        clear_screen()
        print(f"HP: {self.player_hp:3d}   Score: {self.score:4d}")
        print("-" * 40)
        print("Commands: attack | heal | quit")

    def run(self):
        print("=== Simple ASCII Game ===\n")
        time.sleep(1.2)

        while self.running:
            self.render()

            if self.player_hp <= 0:
                print("\n*** GAME OVER ***")
                print(f"Final score: {self.score}")
                self.running = False
                break

            self.handle_input()
            self.update()
            time.sleep(0.15)  # ~60 updates per minute


if __name__ == "__main__":
    game = Game()
    try:
        game.run()
    except KeyboardInterrupt:
        print("\nGame terminated by user.")
```

```python
class Book:
    def __init__(self, title, author="unknown", year=None):
        self.title = title
        self.author = author
        self.year = year

    def __str__(self):
        return f"{self.title} by {self.author}" + (f"{self.year}" if self.year else "")

class Library:
    def __init__(self):
        self.books = []

    @property
    def no_of_books(self):
        return len(self.books)

    def add_book(self, title, author="Unknown", year=None):
        book = Book(title, author, year)
        self.books.append(book)

    def show_info(self):
        print(f"\n📚 The library has {self.no_of_books} book{'s' if self.no_of_books != 1 else ''}.")
        for i, book in enumerate(self.books, 1):
            print(f"   {i}. {book}")

    def remove_book(self, book_title):
        for book in self.books:
            if str(book).lower().startswith(book_title.lower()):
                self.books.remove(book)
                print(f"✅ Removed: {book}")
                return
        print("❌ Book not found!")

    def search_book(self, keyword):
        found = [book for book in self.books if keyword.lower() in str(book).lower()]
        if found:
            print(f"\n🔍 Found {len(found)} book(s) matching '{keyword}':")
            for book in found:
                print(f"   • {book}")
        else:
            print("❌ No matching books found.")

if __name__ == "__main__":

    l1 = Library()
    l1.add_book("Python Crash Course", "Eric Matthes", 2019)
    l1.add_book("Clean Code", "Robert C. Martin")
    l1.add_book("1984", "George Orwell", 1949)
    l1.show_info()

    l1.add_book("Harry Potter")
    l1.add_book("The Alchemist")
    l1.add_book("Dune")
    l1.show_info()
    l1.search_book("harry")
    l1.remove_book("alchemist")
    l1.show_info()
```

```python
def tribonacci(n):
    """
    Generate first n terms of Tribonacci series
    """
    if n <= 0:
        return []
    elif n == 1:
        return [0]
    elif n == 2:
        return [0, 0]
    elif n == 3:
        return [0, 0, 1]

    # Initialize first three terms
    series = [0, 0, 1]

    for i in range(3, n):
        next_term = series[-1] + series[-2] + series[-3]
        series.append(next_term)

    return series

def main():
    print("=== Tribonacci Series Generator ===")
    try:
        n = int(input("Enter number of terms: "))
        if n <= 0:
            print("Please enter a positive number.")
            return

        series = tribonacci(n)
        print(f"\nFirst {n} terms of Tribonacci series:")
        print(series)

        # Print nicely formatted
        print("\nFormatted:")
        for i, num in enumerate(series):
            print(f"T({i}) = {num}")

    except ValueError:
        print("Invalid input! Please enter an integer.")

if __name__ == "__main__":
    main()
```
